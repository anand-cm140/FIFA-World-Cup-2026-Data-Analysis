# 📊 Team Performance Analysis

This section calculates tournament statistics for every team.

The following metrics are computed:

- Matches Played
- Wins
- Draws
- Losses
- Goals Scored
- Goals Conceded
- Goal Difference

These statistics will be used for visualizations later in the notebook.

## Matches

```python
def matches_palyed(team):
    home_matches = (clean_df["home_team"] == team).sum()

    away_matches = (clean_df["away_team"] == team ).sum()
    
    return home_matches +  away_matches
```

```python
teams = teams["teams"].tolist()

matches = []

for team in teams:
    matches.append(matches_palyed(team))

```

```python
print(teams)
```

```python
print(matches)
```

## Matches Won

```python
clean_df.head()
```

```python
def matches_won(team):
    home_win =(
        (clean_df["home_team"] == team) 
        &
        (clean_df["home_score"] > clean_df["away_score"])
    ).sum()

    away_win =(
        (clean_df["away_team"] == team) 
        &
        (clean_df["home_score"] < clean_df["away_score"])
    ).sum()

    return  home_win  +  away_win

    
```

```python
won = []

for team in teams:
    won.append(matches_won(team))
```

```python
won
```

## Matches Lost

```python
def matches_loss(team):
    home_loss =(
        (clean_df["home_team"] == team) 
        &
        (clean_df["home_score"] > clean_df["away_score"])
    ).sum()

    away_loss =(
        (clean_df["away_team"] == team) 
        &
        (clean_df["home_score"] < clean_df["away_score"])
    ).sum()

    return  home_loss  +  away_loss
```

```python
loss = []

for team in teams:
    loss.append(matches_loss(team))
```

## Draw 

```python
clean_df.head()
```

```python
def matches_draw(team):
    home_draw =(
        (clean_df["home_team"] == team) 
        &
        (clean_df["home_score"] == clean_df["away_score"])
    ).sum()

    away_draw =(
        (clean_df["away_team"] == team) 
        &
        (clean_df["home_score"] == clean_df["away_score"])
    ).sum()

    return  home_draw  +  away_draw
```

```python
draw = []

for team in teams:
    draw.append(matches_draw(team))
```

## Goals Scored

```python
clean_df.head()
```

```python
def goals_scored(team):
    home_score =(
        (clean_df["home_team"] == team) 
        &
        (clean_df["home_score"])
    ).sum()

    away_score=(
        (clean_df["away_team"] == team) 
        &
        (clean_df["home_score"])
    ).sum()

    return  home_score  +  away_score
```

```python
goals = []

for team in teams:
    goals.append(goals_scored(team))
```

## Goals Conceded

```python
def goals_conceded(team):
    home_conceded =(
        (clean_df["home_team"] == team) 
        &
        (clean_df["away_score"])
    ).sum()

    away_conceded=(
        (clean_df["away_team"] == team) 
        &
        (clean_df["home_score"])
    ).sum()

    return  home_conceded  +  away_conceded
```

```python
conceded = []

for team in teams:
    conceded.append(goals_conceded(team))
```

## Goal Difference

```python
gd = []

for i in range(len(teams)):
    gd.append(goals [i] - conceded [i])
```

 ## 📋 Team Statistics Summary

```python
team_data_df = pd.DataFrame({
        "Teams" : teams,
        "Matches" : matches,
        "Won": won,
        "Loss": loss,
        "Draw": draw,
        "Goals Scored": goals,
        "Goals Conceded": conceded,
        "GD" : gd
        

})
```

```python
x = team_data_df["Goals Scored"]
y = team_data_df["Won"]

# Create figure
plt.figure(figsize=(10, 6))

# Scatter plot
plt.scatter(
    x,
    y,
    s=80,
    alpha=0.9,
    edgecolors="black",
    linewidths=0.8,
    color="royalblue"
)

# Trend line
m, b = np.polyfit(x, y, 1)
plt.plot(
    x,
    m * x + b,
    color="crimson",
    linewidth=2.5,
    label="Trend Line"
)

# Correlation
corr = x.corr(y)

# Labels
plt.title(
    "Goals Scored vs Matches Won - FIFA World Cup 2026",
    fontsize=15,
    weight="bold"
)
plt.xlabel("Goals Scored", fontsize=12)
plt.ylabel("Matches Won", fontsize=12)

# Grid
plt.grid(alpha=0.3)

# Legend
plt.legend()

# Correlation text
plt.text(
    0.98,
    0.05,
    f"Correlation = {corr:.2f}",
    transform=plt.gca().transAxes,
    ha="right",
    fontsize=11,
    bbox=dict(facecolor="white", edgecolor="black")
)
plt.tight_layout()
plt.savefig('images/scatter_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Insight

- A positive correlation (r = 0.74) exists between goals scored and matches won.
- Teams that scored more goals generally won more matches.
